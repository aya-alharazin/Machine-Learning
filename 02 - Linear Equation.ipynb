```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 100)  # 100 points from -5 to 5
y = 2*x - 6

plt.plot(x, y, color='red', label='y = 2x - 6')
plt.axhline(0, color='black', linewidth=0.5)  # x-axis
plt.axvline(0, color='black', linewidth=0.5)  # y-axis
plt.scatter([3, -2], [0, -10], color='blue', zorder=5)  # our 2 points
plt.title("Graph for y = 2*x - 6")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ── The 3 equations from the slide ──
equations = [
    {"m": 1,  "c": 1, "label": "y = x + 1"},
    {"m": 5,  "c": 1, "label": "y = 5x + 1"},
    {"m": 30, "c": 1, "label": "y = 30x + 1"},
]

# ── Create 3 side-by-side graphs ──
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, eq in zip(axes, equations):
    m = eq["m"]
    c = eq["c"]

    # Generate x values (small range so steep lines still look good)
    x = np.linspace(-0.5, 0.5, 100)
    y = m * x + c

    # Plot the line
    ax.plot(x, y, color='blue', linewidth=2, label=eq["label"])

    # Mark 2 key points
    x1, x2 = 0, 1/m          # when x=0 and when y=0
    y1, y2 = c, 0
    ax.scatter([x1, x2], [y1, y2], color='blue', zorder=5)
    ax.annotate(f'(0, {c})',   (x1, y1), textcoords="offset points", xytext=(5, -15), fontsize=9)
    ax.annotate(f'({x2:.2f}, 0)', (x2, y2), textcoords="offset points", xytext=(5, 5),  fontsize=9)

    # Draw x and y axes
    ax.axhline(0, color='black', linewidth=0.8)
    ax.axvline(0, color='black', linewidth=0.8)

    # Labels and info
    ax.set_title(eq["label"], fontsize=13, fontweight='bold')
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.grid(True, linestyle='--', alpha=0.4)
    ax.legend(fontsize=9)

    # Add slope info as text inside graph
    ax.text(0.05, 0.95,
            f"When x=0, y={c}\nWhen y=0, x={round(-c/m, 4)}",
            transform=ax.transAxes,
            fontsize=8,
            verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))

plt.suptitle("Slope Steepness: The bigger m, the steeper the line",
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# ── Graph 1: Positive Slope ──
ax1 = axes[0]
x1 = [0, 4]
y1 = [1, 5]
ax1.plot(x1, y1, color='blue', linewidth=2.5)
ax1.scatter(x1, y1, color='blue', zorder=5)
ax1.annotate('(0, 1)', (0, 1), textcoords="offset points", xytext=(-30, -15), fontsize=10)
ax1.annotate('(4, 5)', (4, 5), textcoords="offset points", xytext=(5, 5),    fontsize=10)
ax1.text(1.5, 3, '+ve slope', color='red', fontsize=12, fontweight='bold')
ax1.axhline(0, color='black', linewidth=0.8)
ax1.axvline(0, color='black', linewidth=0.8)
ax1.set_title('Positive Slope', fontsize=13, fontweight='bold', color='green')
ax1.set_xlim(-1, 6)
ax1.set_ylim(-1, 7)
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.grid(True, linestyle='--', alpha=0.3)

# Calculate and show slope
m1 = (5 - 1) / (4 - 0)
ax1.text(0.05, 0.05,
         f'm = (5-1)/(4-0) = {m1}',
         transform=ax1.transAxes,
         fontsize=9, color='green',
         bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.4))

# ── Graph 2: Negative Slope ──
ax2 = axes[1]
x2 = [0, 3]
y2 = [5, 1]
ax2.plot(x2, y2, color='blue', linewidth=2.5)
ax2.scatter(x2, y2, color='blue', zorder=5)
ax2.annotate('(0, 5)', (0, 5), textcoords="offset points", xytext=(5, 5),    fontsize=10)
ax2.annotate('(3, 1)', (3, 1), textcoords="offset points", xytext=(5, -15),  fontsize=10)
ax2.text(0.8, 3.5, '-ve slope', color='red', fontsize=12, fontweight='bold')
ax2.axhline(0, color='black', linewidth=0.8)
ax2.axvline(0, color='black', linewidth=0.8)
ax2.set_title('Negative Slope', fontsize=13, fontweight='bold', color='red')
ax2.set_xlim(-1, 5)
ax2.set_ylim(-1, 7)
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.grid(True, linestyle='--', alpha=0.3)

m2 = (1 - 5) / (3 - 0)
ax2.text(0.05, 0.05,
         f'm = (1-5)/(3-0) = {round(m2, 2)}',
         transform=ax2.transAxes,
         fontsize=9, color='red',
         bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.4))

# ── Graph 3: Zero Slope ──
ax3 = axes[2]
x3 = [0, 3]
y3 = [5, 5]
ax3.plot(x3, y3, color='blue', linewidth=2.5)
ax3.scatter(x3, y3, color='blue', zorder=5)
ax3.annotate('(0, 5)', (0, 5), textcoords="offset points", xytext=(-30, 8),  fontsize=10)
ax3.annotate('(3, 5)', (3, 5), textcoords="offset points", xytext=(5, 8),    fontsize=10)
ax3.text(0.8, 3.5, 'zero slope', color='red', fontsize=12, fontweight='bold')
ax3.axhline(0, color='black', linewidth=0.8)
ax3.axvline(0, color='black', linewidth=0.8)
ax3.set_title('Zero Slope', fontsize=13, fontweight='bold', color='gray')
ax3.set_xlim(-1, 5)
ax3.set_ylim(-1, 7)
ax3.set_xlabel('x')
ax3.set_ylabel('y')
ax3.grid(True, linestyle='--', alpha=0.3)

m3 = (5 - 5) / (3 - 0)
ax3.text(0.05, 0.05,
         f'm = (5-5)/(3-0) = {m3} → horizontal line',
         transform=ax3.transAxes,
         fontsize=9, color='gray',
         bbox=dict(boxstyle='round', facecolor='lightgray', alpha=0.4))

plt.suptitle('Interpreting the Slope: Positive, Negative, and Zero',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```
