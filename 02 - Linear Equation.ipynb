```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 100)  # 100 points from -5 to 5
y = 2*x - 6

plt.plot(x, y, color='red', label='y = 2x - 6')
plt.axhline(0, color='black', linewidth=0.5)  # x-axis
plt.axvline(0, color='black', linewidth=0.5)  # y-axis
plt.scatter([3, -2], [0, -10], color='blue', zorder=5)  # our 2 points
plt.title("Graph for y = 2*x - 6")
plt.legend()
plt.grid(True)
plt.show()
```

```python


# ── The 3 equations from the slide ──
equations = [
    {"m": 1,  "c": 1, "label": "y = x + 1"},
    {"m": 5,  "c": 1, "label": "y = 5x + 1"},
    {"m": 30, "c": 1, "label": "y = 30x + 1"},
]

# ── Create 3 side-by-side graphs ──
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, eq in zip(axes, equations):
    m = eq["m"]
    c = eq["c"]

    # Generate x values (small range so steep lines still look good)
    x = np.linspace(-0.5, 0.5, 100)
    y = m * x + c

    # Plot the line
    ax.plot(x, y, color='blue', linewidth=2, label=eq["label"])

    # Mark 2 key points
    x1, x2 = 0, 1/m          # when x=0 and when y=0
    y1, y2 = c, 0
    ax.scatter([x1, x2], [y1, y2], color='blue', zorder=5)
    ax.annotate(f'(0, {c})',   (x1, y1), textcoords="offset points", xytext=(5, -15), fontsize=9)
    ax.annotate(f'({x2:.2f}, 0)', (x2, y2), textcoords="offset points", xytext=(5, 5),  fontsize=9)

    # Draw x and y axes
    ax.axhline(0, color='black', linewidth=0.8)
    ax.axvline(0, color='black', linewidth=0.8)

    # Labels and info
    ax.set_title(eq["label"], fontsize=13, fontweight='bold')
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.grid(True, linestyle='--', alpha=0.4)
    ax.legend(fontsize=9)

    # Add slope info as text inside graph
    ax.text(0.05, 0.95,
            f"When x=0, y={c}\nWhen y=0, x={round(-c/m, 4)}",
            transform=ax.transAxes,
            fontsize=8,
            verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))

plt.suptitle("Slope Steepness: The bigger m, the steeper the line",
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.patches as mpatches
from matplotlib.gridspec import GridSpec
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

x = np.linspace(0.2, 6, 300)
y = (x - 3)**2 + 2

def f(x):  return (x - 3)**2 + 2
def df(x): return 2 * (x - 3)   # slope at any point (derivative)

def draw_tangent(ax, tx, color, label):
    slope = df(tx)
    ty    = f(tx)
    # tangent line: y = slope*(x - tx) + ty
    tx_line = np.linspace(tx - 1.2, tx + 1.2, 100)
    ty_line = slope * (tx_line - tx) + ty
    ax.plot(tx_line, ty_line, color=color, linewidth=2,
            linestyle='--', label=label)
    ax.scatter([tx], [ty], color=color, zorder=5, s=80)
    ax.annotate(f'slope={slope:.1f}', (tx, ty),
                textcoords="offset points", xytext=(8, 8),
                fontsize=9, color=color, fontweight='bold')

# Left: show the curve and what tangent means
ax = axes[0]
ax.plot(x, y, color='blue', linewidth=2.5, label='y = (x−3)² + 2')
draw_tangent(ax, 1.5, 'red',    'tangent at x=1.5 (before min)')
draw_tangent(ax, 3.0, 'green',  'tangent at x=3.0 (minimum)')
draw_tangent(ax, 4.5, 'orange', 'tangent at x=4.5 (after min)')
ax.axhline(0, color='black', linewidth=0.8)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlim(0, 6.5)
ax.set_ylim(-1, 12)
ax.set_title('Tangent lines at different points', fontsize=13, fontweight='bold')
ax.set_xlabel('x'); ax.set_ylabel('y')
ax.legend(fontsize=9); ax.grid(True, alpha=0.3)

# Right: slope values across the curve
ax2 = axes[1]
x_slope = np.linspace(0.5, 5.5, 300)
slopes   = df(x_slope)
colors_s = ['blue' if s < -0.05 else 'red' if s > 0.05 else 'green'
            for s in slopes]
ax2.scatter(x_slope, slopes, c=colors_s, s=8)
ax2.axhline(0, color='black', linewidth=1.5, linestyle='--', label='slope = 0')
ax2.axvline(3, color='green', linewidth=1.5, linestyle=':', label='x=3 (minimum)')
ax2.set_title('Slope value at every point on the curve', fontsize=13, fontweight='bold')
ax2.set_xlabel('x'); ax2.set_ylabel('slope (tangent steepness)')
ax2.legend(fontsize=9); ax2.grid(True, alpha=0.3)

blue_p  = mpatches.Patch(color='blue',  label='Negative slope (going down)')
red_p   = mpatches.Patch(color='red',   label='Positive slope (going up)')
green_p = mpatches.Patch(color='green', label='Zero slope (minimum)')
ax2.legend(handles=[blue_p, red_p, green_p], fontsize=9)

plt.suptitle('Visual 1 — What is a Tangent Line?',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

points   = [1.5,   3.0,    4.5  ]
titles   = ['Before (3,2)\nNegative slope',
            'At (3,2)\nZero slope',
            'After (3,2)\nPositive slope']
colors   = ['blue', 'green', 'red']
bg_colors= ['#E6F1FB', '#EAF3DE', '#FCEBEB']

for i, (tx, title, col, bg) in enumerate(
        zip(points, titles, colors, bg_colors)):

    ax = axes[i]
    ax.set_facecolor(bg)

    ax.plot(x, y, color='steelblue', linewidth=2.5,
            label='y = (x−3)² + 2')

    slope  = df(tx)
    ty     = f(tx)
    tx_ln  = np.linspace(tx - 1.5, tx + 1.5, 100)
    ty_ln  = slope * (tx_ln - tx) + ty
    ax.plot(tx_ln, ty_ln, color=col, linewidth=2.5,
            linestyle='--', label=f'tangent (slope={slope:.1f})')
    ax.scatter([tx], [ty], color=col, zorder=5, s=100)

    ax.axhline(0, color='black', linewidth=0.8)
    ax.axvline(0, color='black', linewidth=0.8)
    ax.set_xlim(0, 6); ax.set_ylim(-1, 12)
    ax.set_title(title, fontsize=12, fontweight='bold', color=col)
    ax.set_xlabel('x'); ax.set_ylabel('y')
    ax.legend(fontsize=9); ax.grid(True, alpha=0.3)

    # Big slope label inside graph
    ax.text(0.5, 0.1, f'slope = {slope:.1f}',
            transform=ax.transAxes,
            fontsize=14, fontweight='bold', color=col,
            ha='center',
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))

plt.suptitle('Visual 2 — Positive, Zero, and Negative Tangent Slopes',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Loss curve (what AI tries to minimize)
x_loss = np.linspace(-3, 3, 300)
y_loss = x_loss**2 + 1

# Left: show the descent steps
ax = axes[0]
ax.plot(x_loss, y_loss, color='purple', linewidth=2.5,
        label='Loss curve (model error)')
ax.set_title('How AI walks downhill using slope',
             fontsize=12, fontweight='bold')

start  = -2.5
lr     = 0.3        # learning rate
pos    = start
steps  = []

for step in range(8):
    slope_here = 2 * pos   # derivative of x²+1
    steps.append((pos, pos**2 + 1, slope_here))
    pos = pos - lr * slope_here    # gradient descent step

    # Draw tangent at this step
    tx_ln = np.linspace(pos - 0.5, pos + 0.5, 50)
    ty_ln = slope_here * (tx_ln - (pos + lr * slope_here)) + \
            (pos + lr * slope_here)**2 + 1
    alpha = max(0.15, 1 - step * 0.1)
    ax.plot(tx_ln, ty_ln, color='orange',
            linewidth=1, alpha=alpha, linestyle='--')

# Draw the path of descent
step_x = [s[0] for s in steps]
step_y = [s[1] for s in steps]
ax.plot(step_x, step_y, 'ro-', markersize=8,
        label='AI steps downhill', zorder=5)

for i, (sx, sy, sl) in enumerate(steps):
    ax.annotate(f'step {i+1}\nslope={sl:.2f}',
                (sx, sy),
                textcoords="offset points",
                xytext=(10, 5),
                fontsize=7, color='darkred')

ax.scatter([0], [1], color='green', s=150, zorder=6,
           marker='*', label='Minimum! slope=0')
ax.axhline(0, color='black', linewidth=0.5)
ax.set_xlabel('model parameter'); ax.set_ylabel('error (loss)')
ax.legend(fontsize=9); ax.grid(True, alpha=0.3)

# Right: slope at each step
ax2 = axes[1]
slopes_ai = [s[2] for s in steps]
ax2.bar(range(1, len(slopes_ai)+1), slopes_ai,
        color=['blue' if s < 0 else 'red' if s > 0.1 else 'green'
               for s in slopes_ai])
ax2.axhline(0, color='black', linewidth=1.5, linestyle='--')
ax2.set_title('Slope at each AI step\n(gets closer to 0 each time)',
              fontsize=12, fontweight='bold')
ax2.set_xlabel('step number')
ax2.set_ylabel('tangent slope')
ax2.grid(True, alpha=0.3, axis='y')

for i, s in enumerate(slopes_ai):
    ax2.text(i + 1, s + (0.1 if s >= 0 else -0.3),
             f'{s:.2f}', ha='center', fontsize=9, fontweight='bold')

plt.suptitle('Visual 3 — Gradient Descent: AI Uses Tangent Slope to Learn',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

days       = np.linspace(0, 14, 300)
# Recovery score: starts at 20, peaks around day 7, levels off
recovery   = -0.5*(days - 7)**2 + 90 + np.sin(days)*2
def drec(d): return -(days[1]-days[0]) + 0  # approximate

# Actual derivative numerically
rec_slope  = np.gradient(recovery, days)

ax = axes[0]
ax.plot(days, recovery, color='blue', linewidth=2.5,
        label='Patient recovery score')

# Show tangent at 3 meaningful days
for day, col, lbl in [(3, 'green', 'Day 3\n(improving fast)'),
                      (7, 'orange','Day 7\n(peak, slope≈0)'),
                      (12,'red',   'Day 12\n(stable)')]:
    idx   = np.argmin(np.abs(days - day))
    slope = rec_slope[idx]
    ry    = recovery[idx]
    tx_ln = np.linspace(day - 1.5, day + 1.5, 50)
    ty_ln = slope * (tx_ln - day) + ry
    ax.plot(tx_ln, ty_ln, color=col,
            linewidth=2, linestyle='--', label=f'{lbl} (slope={slope:.1f})')
    ax.scatter([day], [ry], color=col, s=80, zorder=5)

ax.set_title('Patient Recovery Over Time\n(tangent = how fast improving)',
             fontsize=12, fontweight='bold')
ax.set_xlabel('Day'); ax.set_ylabel('Recovery Score')
ax.legend(fontsize=8); ax.grid(True, alpha=0.3)

# Right: slope (rate of improvement) over time
ax2 = axes[1]
ax2.fill_between(days, rec_slope, 0,
                 where=(rec_slope > 0), color='green', alpha=0.3,
                 label='Improving')
ax2.fill_between(days, rec_slope, 0,
                 where=(rec_slope < 0), color='red', alpha=0.3,
                 label='Declining')
ax2.plot(days, rec_slope, color='black', linewidth=1.5)
ax2.axhline(0, color='black', linewidth=1, linestyle='--')
ax2.set_title('Rate of Recovery at Each Day\n(tangent slope over time)',
              fontsize=12, fontweight='bold')
ax2.set_xlabel('Day'); ax2.set_ylabel('Rate of improvement')
ax2.legend(fontsize=9); ax2.grid(True, alpha=0.3)

plt.suptitle('Visual 4 — Real Life: Tangent Slope = Rate of Change',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```
