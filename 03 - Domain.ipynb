$$
f(x,y) = \frac{\ln(y - x)}{\sqrt{x - y + 1}}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Create grid
x = np.linspace(-5, 5, 400)
y = np.linspace(-5, 5, 400)
X, Y = np.meshgrid(x, y)

# Domain condition
domain = (Y > X) & (Y < X + 1)

# Plot
plt.figure()
plt.contourf(X, Y, domain, levels=1)

# Draw boundary lines
plt.plot(x, x, linestyle='--')        # y = x
plt.plot(x, x + 1, linestyle='--')    # y = x + 1

plt.xlabel('x')
plt.ylabel('y')
plt.title('Domain of f(x, y)')
plt.show()
```

$$
f(x, y) = \frac{xy}{x^2 - y^2}
$$

```python

# Create grid
x = np.linspace(-5, 5, 400)
y = np.linspace(-5, 5, 400)
X, Y = np.meshgrid(x, y)

# Mask: domain condition (denominator != 0)
Z = X**2 - Y**2
domain = Z != 0  # valid points

# Plot valid domain
plt.figure()
plt.scatter(X[domain], Y[domain], s=1)

# Plot forbidden lines
plt.plot(x, x, linestyle='--', label='y = x')
plt.plot(x, -x, linestyle='--', label='y = -x')

plt.title("Domain of f(x,y) = xy / (x^2 - y^2)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

$$
f(x,y,z) = \sqrt{9 - x^2 - y^2 - z^2}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Create sphere using spherical coordinates
u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(0, np.pi, 100)

x = 3 * np.outer(np.cos(u), np.sin(v))
y = 3 * np.outer(np.sin(u), np.sin(v))
z = 3 * np.outer(np.ones(np.size(u)), np.cos(v))

# Plot
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

ax.plot_surface(x, y, z)

ax.set_title("Domain Boundary: x^2 + y^2 + z^2 = 9")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

$$
f(x,y,z) = \frac{\sqrt{4 - x^2 - y^2}}{z - 3}
$$


```python
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

# Create theta and z
theta = np.linspace(0, 2*np.pi, 100)
z = np.linspace(-5, 5, 100)

theta, z = np.meshgrid(theta, z)

# Cylinder radius = 2
x = 2 * np.cos(theta)
y = 2 * np.sin(theta)

# Plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(x, y, z)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Domain: Cylinder x^2 + y^2 ≤ 4')

plt.show()
```

```python
# Create grid for plane z = 3
x_plane = np.linspace(-3, 3, 50)
y_plane = np.linspace(-3, 3, 50)

x_plane, y_plane = np.meshgrid(x_plane, y_plane)
z_plane = 3 * np.ones_like(x_plane)

# Plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Cylinder
theta = np.linspace(0, 2*np.pi, 100)
z = np.linspace(-5, 5, 100)
theta, z = np.meshgrid(theta, z)

x = 2 * np.cos(theta)
y = 2 * np.sin(theta)

ax.plot_surface(x, y, z)

# Plane z = 3
ax.plot_surface(x_plane, y_plane, z_plane)

ax.set_title('Cylinder with forbidden plane z = 3')

plt.show()
```

## Plane

$$
z = 6 - 2x + 3y
$$

```python
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

# Create grid
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

# Compute Z
Z = 6 - 2*X + 3*Y

# Plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Plane: z = 6 - 2x + 3y')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

# Create grid
x = np.linspace(-4, 4, 200)
y = np.linspace(-4, 4, 200)

X, Y = np.meshgrid(x, y)

# Compute Z
Z = 9 - X**2 - Y**2

# Plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Paraboloid: z = 9 - x^2 - y^2')

plt.show()
```
