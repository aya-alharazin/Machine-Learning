$$
f(x,y) = \frac{\ln(y - x)}{\sqrt{x - y + 1}}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Create grid
x = np.linspace(-5, 5, 400)
y = np.linspace(-5, 5, 400)
X, Y = np.meshgrid(x, y)

# Domain condition
domain = (Y > X) & (Y < X + 1)

# Plot
plt.figure()
plt.contourf(X, Y, domain, levels=1)

# Draw boundary lines
plt.plot(x, x, linestyle='--')        # y = x
plt.plot(x, x + 1, linestyle='--')    # y = x + 1

plt.xlabel('x')
plt.ylabel('y')
plt.title('Domain of f(x, y)')
plt.show()
```
